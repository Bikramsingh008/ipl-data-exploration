## IPL Match Data Exploration

In this project, I explored IPL match data (2008-2024) using Python, Pandas, NumPy, and Matplotlib.

**Tools used:** Pandas, NumPy, Matplotlib

**Dataset:** 1095 matches, 20 columns (team1, team2, winner, toss_winner, toss_decision, venue, season, player_of_match, etc.)

**Key findings:**
- Mumbai Indians are the most successful team with 143 wins, followed by Chennai Super Kings (138) and Kolkata Knight Riders (131).
- Winning the toss does NOT reliably predict winning the match — toss winners won only 554 out of 1095 matches (about 50.6%), essentially a coin flip.
- Teams overwhelmingly prefer to field first after winning the toss (704 times) over batting first (391 times), reflecting the modern T20 trend of chasing.
- Eden Gardens is the most-used venue (77 matches), followed by Wankhede Stadium (73).
- AB de Villiers has won the most Player of the Match awards (25), followed by Chris Gayle (22) and Rohit Sharma (19).

**What I learned:**
- How to analyze categorical data using value_counts() and groupby()
- How to create a derived/calculated column (toss_winner_match_winner) to test a hypothesis
- How to use figsize to make charts with many categories readable
- Practiced the same Pandas/NumPy/Matplotlib workflow on a second, different dataset to build confidence
